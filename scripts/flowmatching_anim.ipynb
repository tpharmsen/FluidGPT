```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

# Load data
data = np.load("../slide-deck/anims/xt_trajectory.npz", allow_pickle=True)

# Compute velocity magnitude
data_new = np.sqrt(
    data["xt"][:, :, 0]**2 +
    data["xt"][:, :, 1]**2
)

# Colors matching your slides
BG = "#0a0e14"
TEXT = "#edf2f5"
GRID = "#1b2430"
COOL = "#52d9c4"

# Select 21 frames
frames = data_new[:21, 0, 5]

# Figure setup
fig, ax = plt.subplots(figsize=(6, 5), dpi=150)

fig.patch.set_facecolor(BG)
ax.set_facecolor(BG)

im = ax.imshow(
    frames[0],
    cmap="viridis",
    interpolation="bilinear"
)

ax.axis("off")

title = ax.set_title(
    "Velocity Magnitude",
    color=TEXT,
    fontsize=16,
    pad=12
)

# Colorbar
cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.ax.tick_params(colors=TEXT)
cbar.outline.set_edgecolor(TEXT)

for spine in ax.spines.values():
    spine.set_color(GRID)


def update(i):
    im.set_array(frames[i])
    title.set_text(f"Velocity Magnitude — step {i+1}/21")
    return [im, title]


ani = FuncAnimation(
    fig,
    update,
    frames=len(frames),
    interval=200,
    blit=True
)

plt.close()

HTML(ani.to_jshtml())

fig.patch.set_facecolor('#0a0e14')
fig.patch.set_alpha(1.0)
ani.save(f'../slide-deck/anims/flowmatching_anim.mp4', writer='ffmpeg', fps=5, dpi=150,
    extra_args=[
        '-pix_fmt', 'yuv420p',
        '-vf', 'scale=in_range=full:out_range=full',
        '-color_range', 'pc',
])
```
