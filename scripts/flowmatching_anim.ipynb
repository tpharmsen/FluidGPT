```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

# Load data
data = np.load("../slide-deck/anims/xt_trajectory.npz", allow_pickle=True)

# Compute velocity magnitude
data_new = np.sqrt(
    data["xt"][:, :, 0]**2 +
    data["xt"][:, :, 1]**2
)

# Colors matching your slides
BG = "#0a0e14"

# Select 21 frames
frames = data_new[:21, 0, 5]

# Figure setup — just the image, no axes/title/colorbar
fig, ax = plt.subplots(figsize=(6, 5), dpi=150)
fig.patch.set_facecolor(BG)
ax.set_facecolor(BG)

im = ax.imshow(
    frames[0],
    cmap="viridis",
    interpolation="bilinear",
    vmin=frames[0].min(),
    vmax=frames[0].max()
)

ax.axis("off")
fig.subplots_adjust(left=0, right=1, top=1, bottom=0)


def update(i):
    im.set_array(frames[i])
    im.set_clim(vmin=frames[i].min(), vmax=frames[i].max())
    return [im]


ani = FuncAnimation(
    fig,
    update,
    frames=len(frames),
    interval=200,
    blit=True
)

plt.close()

HTML(ani.to_jshtml())

fig.patch.set_facecolor(BG)
fig.patch.set_alpha(1.0)
ani.save(
    '../slide-deck/anims/flowmatching_anim.mp4',
    writer='ffmpeg',
    fps=5,
    dpi=150,
    extra_args=[
        '-pix_fmt', 'yuv420p',
        '-vf', 'scale=in_range=full:out_range=full',
        '-color_range', 'pc',
    ]
)
```
